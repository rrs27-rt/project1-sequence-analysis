## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Compares an NGS read to a reference to check validity.

    Checks if sequence in uppercase matches the reference.
    If not, specific errors are recorded.

    Args:
        sequence (str): The NGS read being validated.
        reference (str): The known reference that sequence is being
        compared against.

    Returns:
        valid_read (dict): A dictionary indicating the validity of the
            sequence, with the following keys:
            - 'sequence' (str): The sequence in uppercase.
            - 'valid' (bool): True if the sequence exactly
                matches the reference, False otherwise.
            - 'errors' (list): A list of errors found in the
                sequence. This list is empty if there are no errors.
                Possible errors:
                - "wrong_length": sequence and reference length
                    are not the same.
                - "invalid_characters": sequence has characters
                    that are not A/T/C/G."""

    # Normalise sequence: uppercase
    sequence = sequence.upper()

    # Default Values
    valid = True
    errors = []

    # Mismatched sequence and reference lengths cause "wrong_length" error
    if len(sequence) != len(reference):
        valid = False
        errors.append("wrong_length")

    # Invalid bases (not A/T/C/G) cause "invalid_characters" error
    if not all(base in "ATCG" for base in sequence):
        valid = False
        errors.append("invalid_characters")

    # Create Dictionary
    valid_read = {"sequence": sequence, "valid": valid, "errors": errors}

    return valid_read

help(validate_read)
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Identifies mismatched bases and their positions between an
    NGS read and a reference sequence


    Args:
        sequence (str): The NGS read being compared.
        reference (str): The known reference that the sequence
            is being compared against.

    Returns:
        mutations (list): A list of dictionaries that gives the
            positions, reference base, and sequence base of mutations.
            Has the following keys:
            - 'position' (int): The position of the mutation.
            - 'reference_base' (str): The base in the reference sequence.
            - 'sequence_base' (str): The mutated base in the NGS read."""

    # Both sequence and reference are valid, so errors should not be possible

    # Set default mutations list
    # If there are no mutations, the list stays empty
    mutations = []

    # Make dictionary showing what bases do not match and where
    # Then add to the mutations list
    for index, (seq, ref) in enumerate(zip(sequence, reference)):
        if seq != ref:
            mutations.append(
                {
                'position': index + 1,
                'reference_base': ref,
                'read_base': seq
                }
            )

    return mutations


help(find_mutations)
````

````python
# Test your function with this cell

# no mutations
assert find_mutations("ATCG", "ATCG") == []
print(find_mutations("ATCG", "ATCG"))

# 1 mutation at start
assert find_mutations("TTCG", "ATCG") == [
    {'position': 1, 'reference_base': "A", 'read_base': "T"}
]
print(find_mutations("TTCG", "ATCG"))

# 1 mutation at end
assert find_mutations("ATCA", "ATCG") == [
    {'position': 4, 'reference_base': "G", 'read_base': "A"}
]
print(find_mutations("ATCA", "ATCG"))

# non-consecutive mutations at position 1 and 3
assert find_mutations("CTAG", "ATCG") == [
    {'position': 1, 'reference_base': "A", 'read_base': "C"},
    {'position': 3, 'reference_base': "C", 'read_base': "A"}
]
print(find_mutations("CTAG", "ATCG"))

# All mutations
assert find_mutations("TAGC", "ATCG") == [
    {'position': 1, 'reference_base': "A", 'read_base': "T"},
    {'position': 2, 'reference_base': "T", 'read_base': "A"},
    {'position': 3, 'reference_base': "C", 'read_base': "G"},
    {'position': 4, 'reference_base': "G", 'read_base': "C"}
]
print(find_mutations("TAGC", "ATCG"))
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Compares a list of NGS reads against a single reference.

    Iterates through a list of NGS read dictionaries, validates them
        against the reference using validate_read function and if valid,
        identifies mutations using find_mutations function.
        This information is added as a dictionary for each read.
        All dictionaries are then appended to a list, which is returned.

    Args:
        reads (list): A list of dictionaries of NGS reads.
            Has the following keys:
            - 'read_id' (str): The read id of the NGS read.
            - 'sample' (str): The yeast sample number
            - 'group' (str): Sample group "treatment" or "control".
                sample1 - sample3 are the "treatment" group.
                sample4 - sample6 are the "control" group.
            - 'sequence' (str): The DNA sequence of the NGS read.
        reference (str): The known reference the NGS read is compared
            against. Non-empty and only contains uppercase A/T/C/G.

    Returns:
        analsyed_results (list): A list of dictionaries containing the
            validity of all the NGS reads and their mutations.
            Has the following keys:
            - 'read_id' (str): The read id of the NGS read.
            - 'sample' (str): The yest sample number.
            - 'group' (str): The group of the sample, e.g. "treatment".
            - 'sequence' (str): The NGS read sequence in uppercase.
            - 'status' (str): "valid" if no errors. Otherwise "invalid".
            - 'errors' (list): Contains all the errors found. empty if
                no errors found.
                Errors are "wrong_length" and "invalid_characters",
                as seen in validate_read function.
            - 'mutations' (list): List of dictionaries of mutations,
                showing the position, reference base, and mutated read
                base for each mutation in the read.
                Structure of dictionaries is as shown in find_mutations
                function.

    """
    analysed_results = []

    for read in reads:
        # Compare read sequence against reference to test validity
        valid_dict = validate_read(read["sequence"], reference)

        # Use find_mutations if read is valid, and store values in mutation_list
        # Otherwise, make mutation_list empty
        if valid_dict["valid"] == True:
            mutation_list = find_mutations(valid_dict["sequence"], reference)
        else:
            mutation_list = []

        # Make dictionary of read analysis results, and append to a list
        analysed_results.append(
            {
                'read_id': read["read_id"],
                'sample': read["sample"],
                'group': read["group"],
                'sequence': valid_dict["sequence"],
                'status': "valid" if valid_dict["valid"] else "invalid",
                'errors': valid_dict["errors"],
                'mutations': mutation_list
            }
        )

    return analysed_results

help(analyse_reads)
````

````python
# Helper functions
def calculate_reads(analysed_reads):
    """Count valid, invalid and mismatched reads. Then calculates
    mismatch percentage and total mismatch events"""
    valid_reads = 0
    invalid_reads = 0
    mismatch_reads = 0

    # Count valid reads and invalid reads
    for read in analysed_reads:
        if read["status"] == "valid":
            valid_reads +=1
            # If a valid read has a mutation, increment mismatch_reads
            if read["mutations"]:
                mismatch_reads += 1
        elif read["status"] == "invalid":
            invalid_reads += 1

    # Calculate mismatch percentage
    # Set to None if no valid_reads or mismatch_reads
    if valid_reads == 0 or mismatch_reads == 0:
        mismatch_percentage = None
    else:
        percentage = (mismatch_reads / valid_reads) * 100
        mismatch_percentage = round(percentage, 1)

    # Count total mismatch events
    mismatch_events = 0
    for read in analysed_reads:
        mismatch_events += len(read["mutations"])

    return valid_reads, invalid_reads, mismatch_reads, mismatch_percentage, mismatch_events


def count_substitution(analysed_reads):
    """Count the number of nucleotide substitutions e.g. "A -> G" across
    a sample."""
    # If not valid or no mutations, skip to next read
    substitution_counts = {}
    for read in analysed_reads:
        if not read["status"]=="valid":
            continue
        elif not read["mutations"]:
            continue

        # Count substitution types
        for mutation in read["mutations"]:
            key = f'{mutation["reference_base"]} -> {mutation["read_base"]}'
            substitution_counts[key] = substitution_counts.get(key, 0) + 1

    # Sort positions by alphabetical order
    substitution_counts = dict(sorted(substitution_counts.items()))

    return substitution_counts


def count_positions(analysed_reads):
    """Counts the number of mismatches at each positon."""
    position_counts = {}

    # Count mismatch per position
    for read in analysed_reads:
        for mutation in read["mutations"]:
            # Skip if no mismatches
            if not mutation:
                continue

            # Count mismatches at each position
            position = mutation["position"]
            position_counts[position] = position_counts.get(position, 0) + 1

    # sort position_counts by position
    position_counts = dict(sorted(position_counts.items(), key = lambda x:x[0]))
    return position_counts

def top_3_positions(position_counts):
    """"Get the top 3 positions with the most mismatch events in a sample"""
    # Sort positions by number of mismatch events, then by position to account for ties.
    # Turn into list for slicing
    positions_sorted = list(sorted(position_counts.items(), key = lambda x:(-x[1], x[0])))

    # Get top 3 from list, and turn back to a dictionary
    top_positions = dict(positions_sorted[:3])
    return top_positions


# Main function
def summarise_samples(analysed_reads):
    """Output

    Args:
        analysed_reads(list): the output of the analyse_reads function

    Returns:
        sample_dict (dict): Nested dictionary with the read information
            for each sample.
        Has the following keys:
        - group (str): The sample group. "treatment"/"control" in
            sample data.
        - valid_reads (int): The total valid reads in the sample.
        - invalid_reads (int): The total invalid reads in the sample.
        - mismatch_reads (int): The total valid reads with mismatched
            bases in the sample.
        - mismatch_percentage (float or None): The percentage of valid
            reads in the sample that have mismatched reads.
            Returns None if valid_reads or mismatched_reads is 0
        - mismatch_events (int): The total number of mismatched bases
            in the sample.
        - substitusion_counts (dict): A dictionary with all the
            observed base substitutions as a string,
            and their counts as an int. e.g.:
                {"A -> G": 9}
        - position_counts (dict): A dictionary containing the positions
            where base mismatches occurs as an int, and the total times
            a base mismatch occured at that position as an int. e.g.:
                {21: 4} means position 21 had 4 mismatches
        - top_positions (dict): The top 3 positions where the most
            mismatch events occured.
            If a tie occurs, then the earlier position is selected."""
    # Return empty dictionary if input was empty
    if not analysed_reads:
        return {}

    # Sort analysed_reads by sample, then by read_id
    sorted_reads = sorted(analysed_reads, key = lambda r:(r["sample"], r["read_id"]))

    # Separate reads by sample
    grouped_samples = {}
    for read in sorted_reads:
        sample_name = read["sample"]
        if sample_name not in grouped_samples:
            grouped_samples[sample_name] = []
        # Add read to sample list
        grouped_samples[sample_name].append(read)

    # Initialise results dict
    results_dict = {}

    # Loop through each list of read dicitonaries for each sample
    for sample_name, reads_list in grouped_samples.items():
        # Per sample, get the number of valid, invalid and mismatched reads, percentage of valid
        # reads with mismatches, and the total number of mismatch events
        valid_reads, invalid_reads, mismatch_reads, mismatch_percentage, mismatch_events = calculate_reads(reads_list)

        # Calculate substitution number of nucleotide substitutions (eg A>G)
        substitution_counts = count_substitution(reads_list)

        # Get the number of mismatches per position, and the top 3 positions
        position_counts = count_positions(reads_list)
        top_positions = top_3_positions(position_counts)

        # Add to results_dict for each sample
        results_dict[sample_name] = {
            "group": reads_list[0]["group"],
            "valid_reads": valid_reads,
            "invalid_reads": invalid_reads,
            "mismatch_reads": mismatch_reads,
            "mismatch_percentage": mismatch_percentage,
            "mismatch_events": mismatch_events,
            "substitution_counts": substitution_counts,
            "position_counts": position_counts,
            "top_positions": list(top_positions.keys()) # Get list of keys (positions)
        }


    # Table of summarised results
    print("---Table of summarised results---")
    print(f'{"Sample":<8}{"Valid reads":>22}{"Invalid reads":>22}{"Mismatch percentage":>22}{"Mismatch events":>22}{"Top 3 positions":>22}')
    print("-"*118)
    for sample_name, data in results_dict.items():
        print(f'{sample_name:<8}{data["valid_reads"]:>22}{data["invalid_reads"]:>22}{str(data["mismatch_percentage"]):>22}{data["mismatch_events"]:>22}{str(data["top_positions"]):>22}')

    return results_dict

help(summarise_samples)
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

````python
def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Conduct Welch's t-test for the top 3 highly mutated positions between
    control and treated groups.
    Then prints the results of the statistical tests.

    Args:
        sample_summary (dict): the output of the summarise_samples function

    Returns:
        None
    """
    # Get which samples are part of the control or treatment groups
    control_samples = [name for name, data in sample_summary.items() if data["group"] == "control"]
    treatment_samples = [name for name, data in sample_summary.items() if data["group"] == "treatment"]

    total_mismatch_count = {}

    # Count positions for control samples
    for sample_name in control_samples:
        pos_counts = sample_summary[sample_name]["position_counts"]
        for position, count in pos_counts.items():
            if position not in total_mismatch_count:
                total_mismatch_count[position] = 0
            total_mismatch_count[position] += count

    # Count positions for treatment samples
    for sample_name in treatment_samples:
        pos_counts = sample_summary[sample_name]["position_counts"]
        for position, count in pos_counts.items():
            if position not in total_mismatch_count:
                total_mismatch_count[position] = 0
            total_mismatch_count[position] += count

    # Sort by position in ascending order
    total_mismatch_count = dict(sorted(total_mismatch_count.items(), key=lambda x:x[0]))
    # Get top 3 positions
    selected_positions = top_3_positions(total_mismatch_count)

    # Initialise results list
    results = []

    for pos in selected_positions:
        control_percentages = []
        treated_percentages = []

        # Percentages in control samples
        for sample_name in control_samples:
            pos_counts = sample_summary[sample_name]["position_counts"]
            valid_reads = sample_summary[sample_name]["valid_reads"]

            if valid_reads > 0:
                # If position is not found, mismatch_count is 0 instead of skipping
                mismatch_count = pos_counts.get(pos, 0)
                percentage = (mismatch_count / valid_reads) * 100
                control_percentages.append(percentage)

        # Percentages in treatment samples
        for sample_name in treatment_samples:
            pos_counts = sample_summary[sample_name]["position_counts"]
            valid_reads = sample_summary[sample_name]["valid_reads"]

            if valid_reads > 0:
                # If position is not found, mismatch_count is 0 instead of skipping
                mismatch_count = pos_counts.get(pos, 0)
                percentage = (mismatch_count / valid_reads) * 100
                treated_percentages.append(percentage)

        # Calculate means
        control_mean = sum(control_percentages) / len(control_percentages) if control_percentages else None
        treated_mean = sum(treated_percentages) / len(treated_percentages) if treated_percentages else None

        difference = treated_mean - control_mean if (treated_mean is not None and control_mean is not None) else None

        # Stats tests
        stats_results = calculate_statistics(control_percentages, treated_percentages)

        stats_value = stats_results["statistic"]
        p_val = stats_results["p_value"]

        # Add stats results to results list
        results.append({
            "position": pos,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": stats_value,
            "p_value": p_val
        })

    # Print results
    print("\n---Statistical Results---")

    for res in results:
        control_mean_str = f"{res['control_mean']:.1f} %" if res["control_mean"] else "0.0 %"
        treated_mean_str = f"{res['treated_mean']:.1f} %" if res["treated_mean"] else "0.0 %"
        difference_str = f"{res['difference']:.1f} %" if res["difference"] else "N/A"
        statistic_str = f"{res['statistic']:.2f}" if res["statistic"] else "N/A"
        p_value_str = f"{res['p_value']:.5f}" if res["p_value"] else "N/A"
        print(f"Position {res['position']}, Control (n = {res['control_n']}) mean = {control_mean_str}, Treated (n = {res['treated_n']}) mean = {treated_mean_str}, Difference = {difference_str}, Welch's t-test: t = {statistic_str}, p-value = {p_value_str}")
help(compare_positions)
````
